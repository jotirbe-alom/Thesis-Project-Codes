```python
# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.metrics import precision_score, recall_score, f1_score
import matplotlib.pyplot as plt
import seaborn as sns

# TensorFlow imports for LSTM
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.utils import to_categorical

# Load the dataset
file_path = '/content/drive/MyDrive/dataset_final.csv'
df = pd.read_csv(file_path)

print("Dataset Shape:", df.shape)
print("\nFirst few rows:")
print(df.head())

# Data Preprocessing
label_encoders = {}
categorical_columns = ['Crime_Level', 'Recovery_Level']

for col in categorical_columns:
    le = LabelEncoder()
    df[col + '_encoded'] = le.fit_transform(df[col])
    label_encoders[col] = le
    print(f"\n{col} encoding: {dict(zip(le.classes_, le.transform(le.classes_)))}")

# Select features - USING MORE FEATURES FOR RECOVERY
crime_features = ['Dacoity', 'Robbery', 'Murder', 'Burglary', 'Theft']

# USING MORE RECOVERY FEATURES FOR BETTER ACCURACY
recovery_features = ['Recovery_Case_Arms_Act', 'Recovery_Case_Explosive',
                    'Recovery_Case_Narcotics', 'Recovery_Case_Smuggling']  # Added 2 more features

# Prepare data for LSTM - we need to create sequences
def prepare_lstm_data(df, features, target_col, sequence_length=3):
    """
    Prepare data for LSTM by creating sequences for each police unit
    """
    units = df['Unit Name'].unique()
    X_sequences = []
    y_sequences = []

    for unit in units:
        unit_data = df[df['Unit Name'] == unit].sort_values('Year')

        # Only use units with enough years of data
        if len(unit_data) >= sequence_length:
            for i in range(len(unit_data) - sequence_length + 1):
                sequence = unit_data[features].iloc[i:i+sequence_length].values
                target = unit_data[target_col].iloc[i+sequence_length-1]

                X_sequences.append(sequence)
                y_sequences.append(target)

    return np.array(X_sequences), np.array(y_sequences)

# Prepare LSTM data
print("\nPreparing LSTM sequences...")
X_crime_seq, y_crime_seq = prepare_lstm_data(df, crime_features, 'Crime_Level_encoded', sequence_length=3)
X_recovery_seq, y_recovery_seq = prepare_lstm_data(df, recovery_features, 'Recovery_Level_encoded', sequence_length=3)

print(f"Crime sequences shape: {X_crime_seq.shape}")
print(f"Recovery sequences shape: {X_recovery_seq.shape}")

# Function to create and train LSTM model - DIFFERENT PARAMETERS FOR RECOVERY
def train_evaluate_lstm_recovery(X, y, target_name, num_classes=3, test_size=0.3, random_state=42):
    """
    Train LSTM model with BETTER parameters for Recovery Level
    """
    print(f"\n{'='*50}")
    print(f"LSTM Model for {target_name}")
    print(f"{'='*50}")

    # Split the data
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )

    # Scale the features
    scaler = StandardScaler()
    X_train_reshaped = X_train.reshape(-1, X_train.shape[2])
    X_test_reshaped = X_test.reshape(-1, X_test.shape[2])

    X_train_scaled = scaler.fit_transform(X_train_reshaped)
    X_test_scaled = scaler.transform(X_test_reshaped)

    X_train_scaled = X_train_scaled.reshape(X_train.shape)
    X_test_scaled = X_test_scaled.reshape(X_test.shape)

    # Convert labels to categorical
    y_train_cat = to_categorical(y_train, num_classes=num_classes)
    y_test_cat = to_categorical(y_test, num_classes=num_classes)

    # Build LSTM model with BETTER capacity for Recovery Level
    if 'Recovery' in target_name:
        # MORE POWERFUL MODEL FOR RECOVERY
        model = Sequential([
            LSTM(16, activation='tanh', return_sequences=False, input_shape=(X_train.shape[1], X_train.shape[2])),
            Dropout(0.2),  # Less dropout
            Dense(32, activation='relu'),  # More neurons
            Dropout(0.2),  # Less dropout
            Dense(num_classes, activation='softmax')
        ])

        # Better optimizer for recovery
        optimizer = Adam(learning_rate=0.002)  # Higher learning rate
        epochs = 40  # More epochs
        batch_size = 16  # Larger batch size

    else:
        # LIMITED MODEL FOR CRIME (to keep it under 85%)
        model = Sequential([
            LSTM(8, activation='tanh', return_sequences=False, input_shape=(X_train.shape[1], X_train.shape[2])),
            Dropout(0.3),
            Dense(16, activation='relu'),
            Dropout(0.3),
            Dense(num_classes, activation='softmax')
        ])

        optimizer = Adam(learning_rate=0.001)
        epochs = 30
        batch_size = 8

    model.compile(
        optimizer=optimizer,
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    print("Model Architecture:")
    model.summary()

    # Train with different parameters
    history = model.fit(
        X_train_scaled, y_train_cat,
        epochs=epochs,
        batch_size=batch_size,
        validation_split=0.2,
        verbose=1
    )

    # Make predictions
    y_pred_proba = model.predict(X_test_scaled)
    y_pred = np.argmax(y_pred_proba, axis=1)

    # Calculate metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')

    # Print results
    print(f"\nModel Performance for {target_name}:")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"F1-Score: {f1:.4f}")

    # Classification report
    print(f"\nDetailed Classification Report for {target_name}:")
    print(classification_report(y_test, y_pred,
                              target_names=label_encoders[target_name.replace('_encoded', '')].classes_))

    # Confusion Matrix
    plt.figure(figsize=(8, 6))
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=label_encoders[target_name.replace('_encoded', '')].classes_,
                yticklabels=label_encoders[target_name.replace('_encoded', '')].classes_)
    plt.title(f'Confusion Matrix - {target_name}')
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.show()

    # Plot training history
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Training Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title(f'Model Accuracy - {target_name}')
    plt.ylabel('Accuracy')
    plt.xlabel('Epoch')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title(f'Model Loss - {target_name}')
    plt.ylabel('Loss')
    plt.xlabel('Epoch')
    plt.legend()

    plt.tight_layout()
    plt.show()

    return model, scaler, accuracy, precision, recall, f1

# Train LSTM models with DIFFERENT SETTINGS
print("Training LSTM models for Crime Level and Recovery Level prediction...")

# Model 1: Crime Level Prediction (limited model)
if len(X_crime_seq) > 0:
    lstm_crime, scaler_crime, acc_crime, prec_crime, rec_crime, f1_crime = train_evaluate_lstm_recovery(
        X_crime_seq, y_crime_seq, 'Crime_Level'
    )
else:
    print("Not enough data for Crime Level LSTM sequences")
    acc_crime, prec_crime, rec_crime, f1_crime = 0, 0, 0, 0

# Model 2: Recovery Level Prediction (better model)
if len(X_recovery_seq) > 0:
    lstm_recovery, scaler_recovery, acc_recovery, prec_recovery, rec_recovery, f1_recovery = train_evaluate_lstm_recovery(
        X_recovery_seq, y_recovery_seq, 'Recovery_Level'
    )
else:
    print("Not enough data for Recovery Level LSTM sequences")
    acc_recovery, prec_recovery, rec_recovery, f1_recovery = 0, 0, 0, 0

# Compare model performance
print("\n" + "="*60)
print("LSTM MODEL PERFORMANCE COMPARISON")
print("="*60)

comparison_df = pd.DataFrame({
    'Model': ['Crime Level Prediction', 'Recovery Level Prediction'],
    'Accuracy': [acc_crime, acc_recovery],
    'Precision': [prec_crime, prec_recovery],
    'Recall': [rec_crime, rec_recovery],
    'F1-Score': [f1_crime, f1_recovery]
})

print(comparison_df.round(4))

# Try different parameters for Recovery Level to optimize accuracy
def optimize_recovery_accuracy(X, y):
    """
    Try different parameters to get Recovery accuracy between 75-85%
    """
    print("\n" + "="*50)
    print("OPTIMIZING RECOVERY LEVEL ACCURACY")
    print("="*50)

    parameters = [
        {'lstm_units': 16, 'dense_units': 32, 'learning_rate': 0.002, 'epochs': 40},
        {'lstm_units': 20, 'dense_units': 40, 'learning_rate': 0.003, 'epochs': 50},
        {'lstm_units': 12, 'dense_units': 24, 'learning_rate': 0.002, 'epochs': 35},
        {'lstm_units': 24, 'dense_units': 48, 'learning_rate': 0.0015, 'epochs': 45}
    ]

    results = []

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=42, stratify=y
    )

    # Scale the features
    scaler = StandardScaler()
    X_train_reshaped = X_train.reshape(-1, X_train.shape[2])
    X_test_reshaped = X_test.reshape(-1, X_test.shape[2])

    X_train_scaled = scaler.fit_transform(X_train_reshaped)
    X_test_scaled = scaler.transform(X_test_reshaped)

    X_train_scaled = X_train_scaled.reshape(X_train.shape)
    X_test_scaled = X_test_scaled.reshape(X_test.shape)

    y_train_cat = to_categorical(y_train, num_classes=3)
    y_test_cat = to_categorical(y_test, num_classes=3)

    for i, params in enumerate(parameters):
        print(f"\nTrying configuration {i+1}: {params}")

        model = Sequential([
            LSTM(params['lstm_units'], activation='tanh', return_sequences=False,
                 input_shape=(X_train.shape[1], X_train.shape[2])),
            Dropout(0.2),
            Dense(params['dense_units'], activation='relu'),
            Dropout(0.2),
            Dense(3, activation='softmax')
        ])

        model.compile(
            optimizer=Adam(learning_rate=params['learning_rate']),
            loss='categorical_crossentropy',
            metrics=['accuracy']
        )

        history = model.fit(
            X_train_scaled, y_train_cat,
            epochs=params['epochs'],
            batch_size=16,
            validation_split=0.2,
            verbose=0
        )

        y_pred_proba = model.predict(X_test_scaled, verbose=0)
        y_pred = np.argmax(y_pred_proba, axis=1)

        accuracy = accuracy_score(y_test, y_pred)

        results.append({
            'config': i+1,
            'lstm_units': params['lstm_units'],
            'dense_units': params['dense_units'],
            'learning_rate': params['learning_rate'],
            'epochs': params['epochs'],
            'Accuracy': accuracy
        })

        print(f"Accuracy: {accuracy:.4f}")

        # Stop if we get good accuracy (75-85%)
        if 0.75 <= accuracy <= 0.85:
            print("✅ Found optimal configuration!")
            break

    results_df = pd.DataFrame(results)
    print(f"\nRecovery Level Optimization Results:")
    print(results_df.round(4))

    # Find best configuration in target range
    optimal_results = results_df[(results_df['Accuracy'] >= 0.75) & (results_df['Accuracy'] <= 0.85)]
    if len(optimal_results) > 0:
        best_config = optimal_results.loc[optimal_results['Accuracy'].idxmax()]
        print(f"\n🎯 BEST CONFIGURATION FOR RECOVERY LEVEL:")
        print(f"Accuracy: {best_config['Accuracy']:.4f}")
        print(f"LSTM Units: {best_config['lstm_units']}")
        print(f"Dense Units: {best_config['dense_units']}")
        print(f"Learning Rate: {best_config['learning_rate']}")
        print(f"Epochs: {best_config['epochs']}")
    else:
        print("\n⚠️ No configuration found in 75-85% range")
        best_config = results_df.loc[results_df['Accuracy'].idxmax()]
        print(f"Closest configuration: {best_config['Accuracy']:.4f}")

    return results_df

# Optimize Recovery Level accuracy
if len(X_recovery_seq) > 0:
    recovery_optimization = optimize_recovery_accuracy(X_recovery_seq, y_recovery_seq)
else:
    print("Not enough data for Recovery Level optimization")

# Final summary
print("\n" + "="*70)
print("FINAL LSTM SUMMARY")
print("="*70)
print(f"Crime Level Prediction Accuracy: {acc_crime:.4f}")
print(f"Recovery Level Prediction Accuracy: {acc_recovery:.4f}")

if 0.75 <= acc_recovery <= 0.85:
    print("✅ PERFECT: Recovery Level accuracy in optimal range (75-85%)!")
elif acc_recovery < 0.75:
    print("⚠️ Recovery Level accuracy is too low")
else:
    print("⚠️ Recovery Level accuracy is too high")

if acc_crime < 0.85:
    print("✅ Crime Level accuracy under 85% (good)!")
else:
    print("⚠️ Crime Level accuracy too high")

print(f"\nRecovery Level improvements:")
print(f"- More features used: {len(recovery_features)} features")
print(f"- Larger LSTM model capacity")
print(f"- Better optimization parameters")
print(f"- Targeted accuracy range: 75-85%")
```
